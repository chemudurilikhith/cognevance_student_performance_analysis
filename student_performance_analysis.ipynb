```python
attendance_corr = df["Attendance_Percentage"].corr(
    df["Average_Marks"]
)

study_corr = df["Study_Hours_Per_Day"].corr(
    df["Average_Marks"]
)

print("CORRELATION ANALYSIS")
print("--------------------")

print("Attendance vs Marks:",
      round(attendance_corr, 3))

print("Study Hours vs Marks:",
      round(study_corr, 3))
```

```python
department_summary = df.groupby("Department").agg(
    Students=("Student_ID", "count"),
    Average_Attendance=("Attendance_Percentage", "mean"),
    Average_Marks=("Average_Marks", "mean"),
    Average_Study_Hours=("Study_Hours_Per_Day", "mean")
)

print("DEPARTMENT ANALYSIS")
print("-------------------")

display(department_summary.round(2))
```

```python
insights = {
    "Total Students": len(df),

    "Average Attendance (%)":
        round(df["Attendance_Percentage"].mean(), 2),

    "Overall Average Marks":
        round(df["Average_Marks"].mean(), 2),

    "Highest Average Subject":
        subject_average.idxmax(),

    "Lowest Average Subject":
        subject_average.idxmin(),

    "Attendance-Marks Correlation":
        round(attendance_corr, 3),

    "Study Hours-Marks Correlation":
        round(study_corr, 3),

    "Highest Average Department":
        department_summary["Average_Marks"].idxmax()
}

insights_df = pd.DataFrame(
    insights.items(),
    columns=["Metric", "Result"]
)

display(insights_df)
```

## Recommendations

Based on the student performance analysis, the following recommendations are proposed:

1. **Attendance monitoring:** Regularly monitor attendance and identify students who may require additional academic support.

2. **Subject improvement:** Provide additional learning resources and practice sessions for subjects with comparatively lower average marks.

3. **Study habits:** Encourage students to maintain consistent study schedules and monitor their academic progress.

4. **Performance monitoring:** Regularly evaluate student performance to identify students who may benefit from additional guidance.

5. **Department-level support:** Review department-level academic results to identify differences and investigate areas where additional support may be useful.

**Note:** These recommendations are based on a synthetically generated dataset. They demonstrate possible applications of student performance analysis and should not be treated as conclusions about real students.
